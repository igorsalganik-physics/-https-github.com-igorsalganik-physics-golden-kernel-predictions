```python
"""
stage1_patch.py -- corrections to the recovered Stage 1 harness

Apply these three edits to the single-file Stage 1 script before re-running.
Everything else in that file is left untouched, so the run remains a
replication of the registered instrument rather than a new study.

-----------------------------------------------------------------------------
EDIT 1 (REQUIRED) -- the layer-indexing bug in mass_measures()
-----------------------------------------------------------------------------
In mass_measures(), inside `for li, layer in enumerate(attn):`, the line

    mass[:, :, j] = 0.0 if vis.size == 0 else a[:, j, vis].sum(axis=-1)

writes to EVERY layer on every iteration, so each layer overwrites the last
and only the FINAL layer survives. entropies_from_attention() does this
correctly (`H1[li] = ...`), which is why the entropy family is unaffected.

Verified on synthetic attention with known per-layer anchor mass
[0.10, 0.30, 0.50, 0.70, 0.90]: the buggy version reports 0.90 for every
layer; the corrected version recovers all five exactly.

Consequence: every recorded `mass`, `lift` and `Lambda` from a run of the
unpatched file is a final-layer measurement, not a model-wide mean, and
`by_layer_Lambda` is constant across layers. The final layer is where
attention is most peaked, hence where mass sits closest to 0 or 1, hence
where Lambda's boundary derivative 1/[mass(1-mass)] is largest -- so the bug
selects the single worst layer for Lambda and discards the rest. Any prior
finding that Lambda is badly length-distorted must be re-established on the
corrected code before it is reported.

THE FIX: change `mass[:, :, j]` to `mass[li, :, j]`.

-----------------------------------------------------------------------------
EDIT 2 (REQUIRED for section 6) -- record what means cannot recover
-----------------------------------------------------------------------------
The harness records scalar means. Two quantities that the manuscript's
section 6 argument turns on are not recoverable from a mean and must be
recorded at collection time:

  * how close per-head, per-position mass sits to 0 or 1
      -> tests the boundary mechanism behind any Lambda failure
  * the share of attention held by sink positions
      -> tests the null-misspecification mechanism

Replace mass_measures() wholesale with the version below, which fixes EDIT 1
and returns both. Then extend the record in run_item() as shown in EDIT 3.

-----------------------------------------------------------------------------
EDIT 3 (REQUIRED) -- widen the record
-----------------------------------------------------------------------------
In run_item(), replace the `else:` branch of the mass block with the version
at the bottom of this file.

-----------------------------------------------------------------------------
DISCLOSURE
-----------------------------------------------------------------------------
Correcting a demonstrable coding error is a correction to the instrument, not
a change to the analysis plan, so the re-run remains confirmatory under
OSF WQ792. It is disclosable and should be disclosed: state that the original
mass-family code collapsed to the final layer, that it was corrected, and that
the mass-family hypotheses were re-estimated on the corrected code. The
two-sided registration of the Lambda hypothesis was justified by a pilot run
on the uncorrected code; that rationale is weaker than it appeared, but
two-sided remains the conservative choice and needs no change.
"""

import numpy as _np


# ===========================================================================
# EDIT 2 -- corrected and extended mass_measures()
# ===========================================================================
def mass_measures(attn, anchor_idx, min_keys=8, eps=1e-12, n_sink=4):
    """
    Attention mass on a designated token subset, and its length-corrected
    form, computed PER LAYER.

    Returns arrays shaped (layers, heads, positions):
      mass    raw attention mass on the anchor tokens
      lift    mass / (m/n), the uniform-null normalisation -- reported ONLY
              so that its failure is visible; it is NOT a correction
      Lambda  logit(mass) - log(m/(n-m)), the length-invariant statistic
      sink    mass on the first n_sink key positions, whatever their content
      d_bound min(mass, 1-mass), the distance to the unit interval's edge

    sink and d_bound exist to separate the two mechanisms that could make
    Lambda fail on trained attention. They cannot be reconstructed from
    condition means, which is why they are recorded here rather than derived
    later.

    Per query position i, only anchors at or before i are counted, since
    causal masking makes later tokens unreachable.
    """
    if anchor_idx is None or len(anchor_idx) == 0:
        return None
    n_layers = len(attn)
    seq = attn[0].shape[-1]
    positions = _np.arange(seq)
    keep = positions >= (min_keys - 1)
    pos_keep = positions[keep]
    anchors = _np.array(sorted(anchor_idx))

    n_heads = attn[0].shape[0]
    shape = (n_layers, n_heads, pos_keep.size)
    mass = _np.empty(shape)
    sink = _np.empty(shape)

    m_vis = _np.empty(pos_keep.size)
    for j, i in enumerate(pos_keep):
        m_vis[j] = int((anchors <= i).sum())

    for li, layer in enumerate(attn):
        a = _np.asarray(layer, dtype=_np.float64)[:, keep, :]
        a = _np.clip(a, 0.0, None)
        a = a / _np.maximum(a.sum(axis=-1, keepdims=True), eps)
        for j, i in enumerate(pos_keep):
            vis = anchors[anchors <= i]
            # ---- THE FIX: li, not a bare colon ----------------------------
            mass[li, :, j] = (0.0 if vis.size == 0
                              else a[:, j, vis].sum(axis=-1))
            sink[li, :, j] = a[:, j, :min(n_sink, i + 1)].sum(axis=-1)

    n_vis = pos_keep + 1.0
    m_safe = _np.clip(m_vis, 1.0, n_vis - 1.0)[None, None, :]
    n_safe = n_vis[None, None, :]
    mass_c = _np.clip(mass, eps, 1.0 - eps)
    lift = mass / _np.maximum(m_safe / n_safe, eps)
    lam = _np.log(mass_c / (1.0 - mass_c)) - _np.log(m_safe / (n_safe - m_safe))
    d_bound = _np.minimum(mass_c, 1.0 - mass_c)

    return {"mass": mass, "lift": lift, "Lambda": lam,
            "sink": sink, "d_bound": d_bound,
            "positions": pos_keep, "n_anchors_visible": m_vis}


# ===========================================================================
# EDIT 3 -- replacement for the `else:` branch of the mass block in run_item()
# ===========================================================================
EDIT3 = '''
                else:
                    rec.update(
                        mass=float(mm["mass"].mean()),
                        lift=float(mm["lift"].mean()),
                        Lambda=float(mm["Lambda"].mean()),
                        n_anchor_tokens=len(idx),
                        by_layer_mass=[float(v) for v in
                                       mm["mass"].mean(axis=(1, 2))],
                        by_layer_Lambda=[float(v) for v in
                                         mm["Lambda"].mean(axis=(1, 2))],
                        # --- boundary mechanism -------------------------
                        d_bound_mean=float(mm["d_bound"].mean()),
                        d_bound_p01=float(np.quantile(mm["d_bound"], 0.01)),
                        d_bound_p05=float(np.quantile(mm["d_bound"], 0.05)),
                        d_bound_p50=float(np.quantile(mm["d_bound"], 0.50)),
                        frac_within_001=float((mm["d_bound"] < 0.01).mean()),
                        frac_within_005=float((mm["d_bound"] < 0.05).mean()),
                        # Lambda restricted to cells away from the edge:
                        # if the boundary mechanism drives any Lambda
                        # instability, this is stable where the mean is not
                        Lambda_interior=float(
                            mm["Lambda"][mm["d_bound"] > 0.05].mean())
                            if (mm["d_bound"] > 0.05).any() else None,
                        # --- null-misspecification mechanism -------------
                        sink_mass=float(mm["sink"].mean()),
                        sink_mass_p95=float(np.quantile(mm["sink"], 0.95)),
                    )
'''


# ===========================================================================
# SELF-TEST -- run this cell after patching, before committing GPU time
# ===========================================================================
def selftest():
    seq, heads, layers = 48, 3, 5
    anchors = [2, 3, 4, 5]
    f = [0.10, 0.30, 0.50, 0.70, 0.90]
    attn = []
    for L in range(layers):
        a = _np.zeros((heads, seq, seq))
        for i in range(seq):
            va = [k for k in anchors if k <= i]
            vn = [k for k in range(i + 1) if k not in anchors]
            if va and vn:
                a[:, i, va] = f[L] / len(va)
                a[:, i, vn] = (1 - f[L]) / len(vn)
            else:
                a[:, i, :i + 1] = 1.0 / (i + 1)
        attn.append(a)

    mm = mass_measures(attn, anchors, min_keys=8)
    got = mm["mass"].mean(axis=(1, 2))
    ok = _np.allclose(got, f, atol=2e-3)
    print("per-layer anchor mass")
    print("  true     ", [f"{v:.3f}" for v in f])
    print("  recovered", [f"{v:.3f}" for v in got])
    print("  " + ("PASS -- layers are distinct, the fix is in place"
                  if ok else
                  "FAIL -- still collapsing to one layer; the edit did not apply"))

    flat = _np.allclose(got, got[0])
    print("\n  layers identical? " + ("YES -- BUG STILL PRESENT" if flat
                                      else "no -- correct"))
    return ok


if __name__ == "__main__":
    selftest()
```

```python
"""
STAGE 1 -- prompt ablation study.  SINGLE-FILE VERSION 3 (PATCHED).

Preregistered: https://doi.org/10.17605/OSF.IO/WQ792

CHANGES FROM VERSION 2, all in the mass family. Nothing else is touched.
  1. mass_measures() wrote `mass[:, :, j]` inside the layer loop, so every
     layer overwrote the previous one and only the FINAL layer survived.
     Fixed to `mass[li, :, j]`. Entropy was never affected.
  2. mass_measures() now also returns per-cell sink mass and distance to the
     unit interval's boundary, neither recoverable from a mean.
  3. run_item() records those, plus by_layer_mass and Lambda restricted to
     cells away from the boundary.
A self-test on synthetic attention with known per-layer mass runs at import
and aborts if the layer fix is not in effect.

DISCLOSURE: correcting a demonstrable coding error is a correction to the
instrument, not a change to the analysis plan, so this remains confirmatory
under WQ792. State in any report that the original mass-family code collapsed
to the final layer and that the mass hypotheses were re-estimated.

HOW TO RUN
  1. colab.research.google.com -> File -> New notebook
  2. Runtime -> Change runtime type -> T4 GPU -> Save
  3. Paste this ENTIRE file into one cell, press Ctrl+Enter
  4. ~25 minutes. Then download records.jsonl from the folder icon.
"""

import subprocess, sys

def _pip(*pkgs):
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", *pkgs], check=False)

print("=" * 62)
print("  STAGE 1  --  VERSION 3   (float32 + chat template + layer fix)")
print("  If you do not see this banner, you are running an OLD file.")
print("=" * 62)
print("installing dependencies (1-2 minutes)...")
_pip("transformers>=4.44", "torch", "accelerate")
print("done\n")

import os, json, math, re, gc, time, pathlib
import numpy as np
import torch

# ----------------------------------------------------------------------
# CONFIGURATION
# ----------------------------------------------------------------------
MODEL_NAME   = "Qwen/Qwen2.5-1.5B-Instruct"
DEVICE       = "cuda" if torch.cuda.is_available() else "cpu"
DTYPE        = torch.float32   # NOT float16: Qwen2.5 ships bfloat16 and
                               # overflows fp16, giving NaN attention.

MIN_KEYS     = 8
MAX_NEW      = 96
SEED         = 20260902

USE_DRIVE = True    # CHANGED: default True. A wiped VM has already cost one
                    # run of this study; the data must land outside /content.
if USE_DRIVE:
    try:
        from google.colab import drive
        drive.mount("/content/drive")
        OUT_DIR = pathlib.Path("/content/drive/MyDrive/stage1_out")
    except Exception as _e:
        print(f"[warn] Drive mount failed ({_e}); falling back to /content")
        OUT_DIR = pathlib.Path("/content/stage1_out")
else:
    OUT_DIR = pathlib.Path("/content/stage1_out")
OUT_DIR.mkdir(parents=True, exist_ok=True)
RECORDS = OUT_DIR / "records.jsonl"

torch.manual_seed(SEED)
np.random.seed(SEED)
print(f"model  {MODEL_NAME}\ndevice {DEVICE} / {DTYPE}\nout    {OUT_DIR}")
assert DTYPE == torch.float32, "must be float32: fp16 gives NaN on Qwen2.5"
print("dtype check: float32 confirmed")


# ----------------------------------------------------------------------
# 1. Prompt construction and ablation
# ----------------------------------------------------------------------
ABLATION_KEYS = ("constraints", "exemplars", "anchors", "connectives")


def build_prompt(item, drop=()):
    """Assemble a prompt, omitting the components named in `drop`."""
    drop = set(drop)
    for k in drop:
        if k not in ABLATION_KEYS:
            raise ValueError(f"unknown ablation: {k}")
    parts = []
    if "anchors" not in drop:
        parts.append(item["antecedent"])
    if "exemplars" not in drop:
        parts.append(item["exemplar"])
    parts.append(item["body"])
    if "constraints" not in drop:
        parts.append(item["constraint"])
    text = " ".join(p.strip() for p in parts if p and p.strip())
    if "connectives" in drop:
        text = strip_connectives(text)
    return text


CONNECTIVES = {
    "the", "a", "an", "that", "which", "who", "and", "or", "but", "so",
    "then", "therefore", "however", "in", "of", "to", "for", "with", "as",
    "is", "are", "was", "were", "be", "been", "please", "kindly", "should",
    "would", "could", "must", "may", "might", "will", "shall", "do", "does",
}


def strip_connectives(text):
    """Telegraphic rewrite: drop function words and sentence punctuation."""
    tokens = re.findall(r"[\w'-]+|[.,;:!?]", text)
    kept = [t for t in tokens
            if t.lower() not in CONNECTIVES and t not in ".,;:!?"]
    return " ".join(kept)


# ----------------------------------------------------------------------
# 2. Length matching
# ----------------------------------------------------------------------
FILLER_NEUTRAL = (
    "This section contains background material retained for reference. "
    "It records routine administrative detail and has no bearing on the task. "
)


def pad_to_token_length(text, target_len, tokenize_fn, filler=FILLER_NEUTRAL,
                        max_rounds=200):
    """Append inert filler until the tokenised length reaches target_len."""
    cur = len(tokenize_fn(text))
    if cur >= target_len:
        return text, cur
    padded = text
    for _ in range(max_rounds):
        if len(tokenize_fn(padded + " " + filler)) > target_len:
            break
        padded = padded + " " + filler
        if len(tokenize_fn(padded)) >= target_len:
            break
    words = filler.split()
    i = 0
    while i < len(words) and len(tokenize_fn(
            padded + " " + " ".join(words[:i + 1]))) <= target_len:
        i += 1
    if i > 0:
        padded = padded + " " + " ".join(words[:i])
    return padded, len(tokenize_fn(padded))


# ----------------------------------------------------------------------
# 3. Attention entropy   (UNCHANGED -- this family was never affected)
# ----------------------------------------------------------------------
def entropies_from_attention(attn, min_keys=8, eps=1e-12):
    """Per-layer, per-head, per-position entropies."""
    n_layers = len(attn)
    seq = attn[0].shape[-1]
    positions = np.arange(seq)
    keep = positions >= (min_keys - 1)
    pos_keep = positions[keep]
    if pos_keep.size == 0:
        raise ValueError(f"sequence of length {seq} has no position with "
                         f"at least {min_keys} keys")
    n_heads = attn[0].shape[0]
    H1 = np.empty((n_layers, n_heads, pos_keep.size))
    H2 = np.empty_like(H1)
    for li, layer in enumerate(attn):
        a = np.asarray(layer, dtype=np.float64)[:, keep, :]
        a = np.clip(a, 0.0, None)
        a = a / np.maximum(a.sum(axis=-1, keepdims=True), eps)
        with np.errstate(divide="ignore", invalid="ignore"):
            loga = np.where(a > eps, np.log(np.maximum(a, eps)), 0.0)
        H1[li] = -(a * loga).sum(axis=-1)
        H2[li] = -np.log(np.maximum((a ** 2).sum(axis=-1), eps))
    ceiling = np.log(pos_keep + 1.0)[None, None, :]
    return {"H1": H1, "H2": H2, "Hnorm": H1 / ceiling,
            "span": np.exp(H1), "positions": pos_keep}


def summarise_entropies(ent, by="global"):
    axes = {"global": (0, 1, 2), "layer": (1, 2), "position": (0, 1)}[by]
    return {k: ent[k].mean(axis=axes)
            for k in ("H1", "H2", "Hnorm", "span")}


# ----------------------------------------------------------------------
# 4. Correctness scoring, against the FULL specification
# ----------------------------------------------------------------------
def score_output(item, text):
    low = text.lower()
    return {
        "constraint_ok": item["forbidden"].lower() not in low,
        "referent_ok": item["referent"].lower() in low,
        "format_ok": bool(re.search(item["format_re"], text.strip(),
                                    flags=re.IGNORECASE | re.DOTALL)),
    }


def overall_correct(scores):
    return all(scores.values())


# ----------------------------------------------------------------------
# 5. Paired analysis
# ----------------------------------------------------------------------
def paired_stats(baseline, ablated):
    b = np.asarray(baseline, dtype=float)
    a = np.asarray(ablated, dtype=float)
    if b.shape != a.shape:
        raise ValueError("baseline and ablated must be paired elementwise")
    d = a - b
    n = d.size
    if n < 2:
        raise ValueError("need at least two pairs")
    mean = float(d.mean())
    sd = float(d.std(ddof=1))
    se = sd / math.sqrt(n)
    return {"n": n, "mean_diff": mean, "sd_diff": sd, "se": se,
            "t": float("inf") if se == 0 else mean / se,
            "dz": float("inf") if sd == 0 else mean / sd,
            "ci95": (mean - 1.96 * se, mean + 1.96 * se)}


def required_n(dz, power=0.80, alpha=0.05):
    z_a, z_b = 1.959963985, 0.841621234
    if dz == 0 or dz != dz:
        return float("inf")
    return math.ceil(((z_a + z_b) / abs(dz)) ** 2)


# ----------------------------------------------------------------------
# 6. Mass-family measures                       *** PATCHED SECTION ***
# ----------------------------------------------------------------------
def build_prompt_with_spans(item, drop=()):
    """As build_prompt, but also returns the character span of each
    surviving component. Spans are what let the mass family be computed at
    all: a mass measure needs to know WHICH tokens are the anchors."""
    drop = set(drop)
    parts, spans, pos = [], {}, 0

    def add(name, text):
        nonlocal pos
        if not text or not text.strip():
            return
        if parts:
            pos += 1                      # the joining space
        s = pos
        parts.append(text.strip())
        pos += len(text.strip())
        spans[name] = (s, pos)

    if "anchors" not in drop:
        add("antecedent", item["antecedent"])
    if "exemplars" not in drop:
        add("exemplar", item["exemplar"])
    add("body", item["body"])
    if "constraints" not in drop:
        add("constraint", item["constraint"])

    text = " ".join(parts)
    if "connectives" in drop:
        return strip_connectives(text), None
    return text, spans


def token_indices_for_spans(offsets, spans, names):
    """Map character spans to token indices via the offset mapping."""
    if spans is None:
        return None
    wanted = [spans[n] for n in names if n in spans]
    if not wanted:
        return []
    idx = []
    for i, (a, b) in enumerate(offsets):
        if a == b:                         # special tokens carry empty spans
            continue
        if any(a < e and b > s for s, e in wanted):
            idx.append(i)
    return idx


def mass_measures(attn, anchor_idx, min_keys=8, eps=1e-12, n_sink=4):
    """Attention mass on a designated token subset, computed PER LAYER.

    Returns arrays shaped (layers, heads, positions):
      mass    raw attention mass on the anchor tokens
      lift    mass / (m/n) -- reported ONLY so its failure is visible;
              it is NOT a correction
      Lambda  logit(mass) - log(m/(n-m)), the length-invariant statistic
      sink    mass on the first n_sink key positions, whatever their content
      d_bound min(mass, 1-mass), distance to the unit interval's edge

    sink and d_bound separate the two mechanisms that could make Lambda fail
    on trained attention; neither is reconstructible from condition means.
    """
    if anchor_idx is None or len(anchor_idx) == 0:
        return None
    n_layers = len(attn)
    seq = attn[0].shape[-1]
    positions = np.arange(seq)
    keep = positions >= (min_keys - 1)
    pos_keep = positions[keep]
    if pos_keep.size == 0:
        return None
    anchors = np.array(sorted(anchor_idx))

    n_heads = attn[0].shape[0]
    shape = (n_layers, n_heads, pos_keep.size)
    mass = np.empty(shape)
    sink = np.empty(shape)

    m_vis = np.empty(pos_keep.size)
    for j, i in enumerate(pos_keep):
        m_vis[j] = int((anchors <= i).sum())

    for li, layer in enumerate(attn):
        a = np.asarray(layer, dtype=np.float64)[:, keep, :]
        a = np.clip(a, 0.0, None)
        a = a / np.maximum(a.sum(axis=-1, keepdims=True), eps)
        for j, i in enumerate(pos_keep):
            vis = anchors[anchors <= i]
            # ---- THE FIX: li, not a bare colon -------------------------
            mass[li, :, j] = (0.0 if vis.size == 0
                              else a[:, j, vis].sum(axis=-1))
            sink[li, :, j] = a[:, j, :min(n_sink, i + 1)].sum(axis=-1)

    n_vis = pos_keep + 1.0
    m_safe = np.clip(m_vis, 1.0, n_vis - 1.0)[None, None, :]
    n_safe = n_vis[None, None, :]
    mass_c = np.clip(mass, eps, 1.0 - eps)
    lift = mass / np.maximum(m_safe / n_safe, eps)
    lam = np.log(mass_c / (1.0 - mass_c)) - np.log(m_safe / (n_safe - m_safe))
    d_bound = np.minimum(mass_c, 1.0 - mass_c)

    return {"mass": mass, "lift": lift, "Lambda": lam,
            "sink": sink, "d_bound": d_bound,
            "positions": pos_keep, "n_anchors_visible": m_vis}


# ----------------------------------------------------------------------
# SELF-TESTS -- abort before GPU time if any estimator is wrong
# ----------------------------------------------------------------------
def _causal_uniform(seq, heads=2, layers=3):
    out = []
    for _ in range(layers):
        a = np.zeros((heads, seq, seq))
        for i in range(seq):
            a[:, i, :i + 1] = 1.0 / (i + 1)
        out.append(a)
    return out


def _causal_onehot(seq, heads=2, layers=3):
    a = np.zeros((heads, seq, seq))
    a[:, :, 0] = 1.0
    return [a.copy() for _ in range(layers)]


_e = entropies_from_attention(_causal_uniform(64), min_keys=MIN_KEYS)
assert np.allclose(_e["H1"], np.log(_e["positions"] + 1.0)[None, None, :], atol=1e-10)
assert np.allclose(_e["Hnorm"], 1.0, atol=1e-10)
assert np.allclose(_e["H2"], _e["H1"], atol=1e-10)

_z = entropies_from_attention(_causal_onehot(64), min_keys=MIN_KEYS)
assert np.allclose(_z["H1"], 0.0, atol=1e-10)
assert np.allclose(_z["Hnorm"], 0.0, atol=1e-10)

_p = summarise_entropies(_e, by="position")
assert np.all(np.diff(_p["H1"]) > 0), "raw entropy must rise with position"
assert np.ptp(_p["Hnorm"]) < 1e-10, "normalised entropy must be flat with position"
assert required_n(0.5) == 32 and required_n(0.8) == 13
print("self-test passed: entropy estimator verified against closed forms")


def _mass_layer_selftest():
    """Synthetic attention with KNOWN, DIFFERENT per-layer anchor mass.
    The pre-patch code returns the final layer's value for every layer."""
    seq, heads = 48, 3
    anchors = [2, 3, 4, 5]
    f = [0.10, 0.30, 0.50, 0.70, 0.90]
    attn = []
    for L in range(len(f)):
        a = np.zeros((heads, seq, seq))
        for i in range(seq):
            va = [k for k in anchors if k <= i]
            vn = [k for k in range(i + 1) if k not in anchors]
            if va and vn:
                a[:, i, va] = f[L] / len(va)
                a[:, i, vn] = (1 - f[L]) / len(vn)
            else:
                a[:, i, :i + 1] = 1.0 / (i + 1)
        attn.append(a)
    got = mass_measures(attn, anchors, min_keys=MIN_KEYS)["mass"].mean(axis=(1, 2))
    assert not np.allclose(got, got[0]), (
        "MASS LAYER BUG STILL PRESENT: every layer reports the same value. "
        "mass_measures must assign mass[li, :, j], not mass[:, :, j].")
    assert np.allclose(got, f, atol=2e-3), (
        f"per-layer mass wrong: got {got}, expected {f}")
    print("self-test passed: mass estimator recovers per-layer values "
          f"{[round(v, 3) for v in got]}")


_mass_layer_selftest()


ITEMS = [
 # ---- documents and publications -------------------------------------
 dict(id="d01", antecedent="The Meridian Report is a quarterly publication produced by the Halvorsen Institute.",
   exemplar="Example: 'Summary: a quarterly review of regional shipping volumes.'",
   body="Summarise it in one sentence, beginning with 'Summary:'.",
   constraint="Do not mention the Halvorsen Institute.",
   forbidden="Halvorsen", referent="Meridian", format_re=r"^Summary:"),
 dict(id="d02", antecedent="The Lindqvist Register lists every lighthouse on the western approaches.",
   exemplar="Example: 'Contains: position and light characteristic for each entry.'",
   body="State what it contains, beginning with 'Contains:'.",
   constraint="Do not use the word 'lighthouse'.",
   forbidden="lighthouse", referent="Lindqvist", format_re=r"^Contains:"),
 dict(id="d03", antecedent="The Ashworth Protocol governs data retention at municipal archives.",
   exemplar="Example: 'Purpose: to standardise how long records are kept.'",
   body="Describe its purpose in one sentence, beginning with 'Purpose:'.",
   constraint="Do not use the word 'archive'.",
   forbidden="archive", referent="Ashworth", format_re=r"^Purpose:"),
 dict(id="d04", antecedent="The Venn Accord settled a long dispute over fishing rights in the strait.",
   exemplar="Example: 'Settled: competing claims to a shared waterway.'",
   body="State what it settled, beginning with 'Settled:'.",
   constraint="Do not use the word 'fishing'.",
   forbidden="fishing", referent="Venn", format_re=r"^Settled:"),
 dict(id="d05", antecedent="The Thorne Codex was compiled by monks over three generations.",
   exemplar="Example: 'Origin: assembled gradually by a religious community.'",
   body="Describe its origin, beginning with 'Origin:'.",
   constraint="Do not mention monks.",
   forbidden="monk", referent="Thorne", format_re=r"^Origin:"),
 dict(id="d06", antecedent="The Calderon Index ranks coastal towns by winter storm exposure.",
   exemplar="Example: 'Ranks: settlements by their vulnerability to weather.'",
   body="State what it ranks, beginning with 'Ranks:'.",
   constraint="Do not mention storms.",
   forbidden="storm", referent="Calderon", format_re=r"^Ranks:"),
 dict(id="d07", antecedent="The Pellow Manuscript describes medieval glassmaking in the Rhine valley.",
   exemplar="Example: 'Describes: techniques used by early craftsmen.'",
   body="State what it describes, beginning with 'Describes:'.",
   constraint="Do not mention the Rhine.",
   forbidden="Rhine", referent="Pellow", format_re=r"^Describes:"),
 dict(id="d08", antecedent="The Orsini Ledger records grain prices across four centuries.",
   exemplar="Example: 'Records: commodity costs over a long period.'",
   body="State what it records, beginning with 'Records:'.",
   constraint="Do not use the word 'grain'.",
   forbidden="grain", referent="Orsini", format_re=r"^Records:"),
 dict(id="d09", antecedent="The Barrowfield Survey mapped every hedgerow in the parish.",
   exemplar="Example: 'Mapped: boundary features across a rural district.'",
   body="State what it mapped, beginning with 'Mapped:'.",
   constraint="Do not use the word 'hedgerow'.",
   forbidden="hedgerow", referent="Barrowfield", format_re=r"^Mapped:"),
 dict(id="d10", antecedent="The Quillon Charter granted market rights to seven villages.",
   exemplar="Example: 'Granted: trading privileges to several settlements.'",
   body="State what it granted, beginning with 'Granted:'.",
   constraint="Do not mention villages.",
   forbidden="village", referent="Quillon", format_re=r"^Granted:"),

 # ---- places ----------------------------------------------------------
 dict(id="p01", antecedent="Kestrel Lake sits in the Calder Basin and freezes over each December.",
   exemplar="Example: 'Fact: the water covers roughly forty square kilometres.'",
   body="State one fact about it, beginning with 'Fact:'.",
   constraint="Do not mention the Calder Basin.",
   forbidden="Calder", referent="Kestrel", format_re=r"^Fact:"),
 dict(id="p02", antecedent="Harrow Fen was drained in the eighteenth century for arable farming.",
   exemplar="Example: 'History: converted to dry land over several decades.'",
   body="Describe its history in one sentence, beginning with 'History:'.",
   constraint="Do not mention farming or farmland.",
   forbidden="farm", referent="Harrow", format_re=r"^History:"),
 dict(id="p03", antecedent="Bramwell Hall houses the county's collection of botanical illustrations.",
   exemplar="Example: 'Holding: several hundred hand-coloured plates.'",
   body="Describe what it holds, beginning with 'Holding:'.",
   constraint="Do not use the word 'county'.",
   forbidden="county", referent="Bramwell", format_re=r"^Holding:"),
 dict(id="p04", antecedent="Ferrant Station was the first stop on the old Northline railway.",
   exemplar="Example: 'Note: the platform was rebuilt in 1954.'",
   body="Give one detail about it, beginning with 'Note:'.",
   constraint="Do not mention the Northline.",
   forbidden="Northline", referent="Ferrant", format_re=r"^Note:"),
 dict(id="p05", antecedent="Caldrey Priory kept a continuous weather record from 1608.",
   exemplar="Example: 'Record: daily observations maintained by the community.'",
   body="Describe the record it kept, beginning with 'Record:'.",
   constraint="Do not mention the year 1608.",
   forbidden="1608", referent="Caldrey", format_re=r"^Record:"),
 dict(id="p06", antecedent="Ravenscar Point is the windiest headland on the eastern coast.",
   exemplar="Example: 'Feature: an exposed promontory with severe conditions.'",
   body="Describe its main feature, beginning with 'Feature:'.",
   constraint="Do not use the word 'coast'.",
   forbidden="coast", referent="Ravenscar", format_re=r"^Feature:"),
 dict(id="p07", antecedent="Ilminster Weir was built to power a woollen mill downstream.",
   exemplar="Example: 'Built for: driving machinery further along the river.'",
   body="State why it was built, beginning with 'Built for:'.",
   constraint="Do not mention wool or woollen.",
   forbidden="wool", referent="Ilminster", format_re=r"^Built for:"),
 dict(id="p08", antecedent="Greymoor Common was enclosed in 1802 and later restored to public use.",
   exemplar="Example: 'Status: returned to open access after a period of private holding.'",
   body="Describe its current status, beginning with 'Status:'.",
   constraint="Do not mention the year 1802.",
   forbidden="1802", referent="Greymoor", format_re=r"^Status:"),
 dict(id="p09", antecedent="Saltwick Bay is known for the fossil beds exposed at low tide.",
   exemplar="Example: 'Known for: remains visible when the water retreats.'",
   body="State what it is known for, beginning with 'Known for:'.",
   constraint="Do not use the word 'fossil'.",
   forbidden="fossil", referent="Saltwick", format_re=r"^Known for:"),
 dict(id="p10", antecedent="Denbury Ridge carries the boundary between two ancient dioceses.",
   exemplar="Example: 'Carries: a long-standing administrative division.'",
   body="State what it carries, beginning with 'Carries:'.",
   constraint="Do not use the word 'diocese'.",
   forbidden="diocese", referent="Denbury", format_re=r"^Carries:"),

 # ---- institutions and organisations ----------------------------------
 dict(id="i01", antecedent="Orsett Works manufactured cast-iron fittings until its closure in 1971.",
   exemplar="Example: 'Product: fittings for domestic plumbing.'",
   body="Name what it produced, beginning with 'Product:'.",
   constraint="Do not mention the year 1971.",
   forbidden="1971", referent="Orsett", format_re=r"^Product:"),
 dict(id="i02", antecedent="The Marlowe Trust funds restoration of disused canal locks.",
   exemplar="Example: 'Funds: repair of abandoned waterway structures.'",
   body="State what it funds, beginning with 'Funds:'.",
   constraint="Do not use the word 'canal'.",
   forbidden="canal", referent="Marlowe", format_re=r"^Funds:"),
 dict(id="i03", antecedent="Hessleton Brewery supplied every public house in the borough.",
   exemplar="Example: 'Supplied: drinking establishments across the district.'",
   body="State whom it supplied, beginning with 'Supplied:'.",
   constraint="Do not use the word 'borough'.",
   forbidden="borough", referent="Hessleton", format_re=r"^Supplied:"),
 dict(id="i04", antecedent="The Trevanion Society catalogues regional dialect recordings.",
   exemplar="Example: 'Catalogues: audio material documenting local speech.'",
   body="State what it catalogues, beginning with 'Catalogues:'.",
   constraint="Do not use the word 'dialect'.",
   forbidden="dialect", referent="Trevanion", format_re=r"^Catalogues:"),
 dict(id="i05", antecedent="Penhale Yard repaired fishing smacks for over two centuries.",
   exemplar="Example: 'Repaired: small working vessels, over a long period.'",
   body="State what it repaired, beginning with 'Repaired:'.",
   constraint="Do not use the word 'smack'.",
   forbidden="smack", referent="Penhale", format_re=r"^Repaired:"),
 dict(id="i06", antecedent="The Aldreth Foundation awards grants to rural libraries.",
   exemplar="Example: 'Awards: funding to small community institutions.'",
   body="State what it awards, beginning with 'Awards:'.",
   constraint="Do not use the word 'library' or 'libraries'.",
   forbidden="librar", referent="Aldreth", format_re=r"^Awards:"),
 dict(id="i07", antecedent="Strathmere Colliery employed most of the men in the valley.",
   exemplar="Example: 'Employed: a large share of the local workforce.'",
   body="State whom it employed, beginning with 'Employed:'.",
   constraint="Do not use the word 'valley'.",
   forbidden="valley", referent="Strathmere", format_re=r"^Employed:"),
 dict(id="i08", antecedent="The Duckworth Bequest established a chair in maritime history.",
   exemplar="Example: 'Established: a senior academic post in a historical field.'",
   body="State what it established, beginning with 'Established:'.",
   constraint="Do not use the word 'maritime'.",
   forbidden="maritime", referent="Duckworth", format_re=r"^Established:"),
 dict(id="i09", antecedent="Wrayburn Mill ground flour for the garrison during the siege.",
   exemplar="Example: 'Produced: provisions for a besieged population.'",
   body="State what it produced, beginning with 'Produced:'.",
   constraint="Do not mention the garrison.",
   forbidden="garrison", referent="Wrayburn", format_re=r"^Produced:"),
 dict(id="i10", antecedent="The Ennismore Board oversees drainage across the low-lying levels.",
   exemplar="Example: 'Oversees: water management in a flat region.'",
   body="State what it oversees, beginning with 'Oversees:'.",
   constraint="Do not use the word 'drainage'.",
   forbidden="drainage", referent="Ennismore", format_re=r"^Oversees:"),

 # ---- methods, processes and instruments -------------------------------
 dict(id="m01", antecedent="The Solvig Method is a technique for tuning stringed instruments.",
   exemplar="Example: 'Use: applied shortly before a public performance.'",
   body="State when it is used, beginning with 'Use:'.",
   constraint="Do not use the word 'stringed'.",
   forbidden="stringed", referent="Solvig", format_re=r"^Use:"),
 dict(id="m02", antecedent="The Talvik Index measures how coastal water temperature shifts between summer and winter.",
   exemplar="Example: 'Measures: the spread between readings taken at different times of year.'",
   body="State what it measures, beginning with 'Measures:'.",
   constraint="Do not use the word 'winter'.",
   forbidden="winter", referent="Talvik", format_re=r"^Measures:"),
 dict(id="m03", antecedent="The Nadeau Procedure stabilises fragile paper before conservation work.",
   exemplar="Example: 'Applied to: delicate material prior to treatment.'",
   body="State what it is applied to, beginning with 'Applied to:'.",
   constraint="Do not use the word 'conservation'.",
   forbidden="conservation", referent="Nadeau", format_re=r"^Applied to:"),
 dict(id="m04", antecedent="The Brigstock Gauge estimates peat depth without excavation.",
   exemplar="Example: 'Estimates: subsurface thickness by indirect means.'",
   body="State what it estimates, beginning with 'Estimates:'.",
   constraint="Do not use the word 'peat'.",
   forbidden="peat", referent="Brigstock", format_re=r"^Estimates:"),
 dict(id="m05", antecedent="The Halloran Scale grades the legibility of worn inscriptions.",
   exemplar="Example: 'Grades: how readable a damaged surface remains.'",
   body="State what it grades, beginning with 'Grades:'.",
   constraint="Do not use the word 'inscription'.",
   forbidden="inscription", referent="Halloran", format_re=r"^Grades:"),
 dict(id="m06", antecedent="The Vestergaard Test detects adulteration in imported honey.",
   exemplar="Example: 'Detects: contamination in a foodstuff.'",
   body="State what it detects, beginning with 'Detects:'.",
   constraint="Do not use the word 'honey'.",
   forbidden="honey", referent="Vestergaard", format_re=r"^Detects:"),
 dict(id="m07", antecedent="The Crowhurst Cycle describes how heathland recovers after burning.",
   exemplar="Example: 'Describes: regeneration of vegetation after fire.'",
   body="State what it describes, beginning with 'Describes:'.",
   constraint="Do not use the word 'heathland'.",
   forbidden="heathland", referent="Crowhurst", format_re=r"^Describes:"),
 dict(id="m08", antecedent="The Immerman Filter removes silt from irrigation channels.",
   exemplar="Example: 'Removes: suspended material from flowing water.'",
   body="State what it removes, beginning with 'Removes:'.",
   constraint="Do not use the word 'irrigation'.",
   forbidden="irrigation", referent="Immerman", format_re=r"^Removes:"),
 dict(id="m09", antecedent="The Rossiter Sequence orders excavation layers by pottery type.",
   exemplar="Example: 'Orders: stratigraphic levels by their contents.'",
   body="State what it orders, beginning with 'Orders:'.",
   constraint="Do not use the word 'pottery'.",
   forbidden="pottery", referent="Rossiter", format_re=r"^Orders:"),
 dict(id="m10", antecedent="The Langmere Assay identifies lead content in old window glass.",
   exemplar="Example: 'Identifies: a metallic component of a historic material.'",
   body="State what it identifies, beginning with 'Identifies:'.",
   constraint="Do not use the word 'lead'.",
   forbidden="lead", referent="Langmere", format_re=r"^Identifies:"),

 # ---- events and periods ------------------------------------------------
 dict(id="e01", antecedent="The Ashbourne Flood displaced two thousand residents in a single night.",
   exemplar="Example: 'Effect: a large population was forced to leave quickly.'",
   body="State its effect, beginning with 'Effect:'.",
   constraint="Do not use the word 'residents'.",
   forbidden="resident", referent="Ashbourne", format_re=r"^Effect:"),
 dict(id="e02", antecedent="The Threlkeld Dispute halted quarry work for eleven months.",
   exemplar="Example: 'Consequence: extraction ceased for nearly a year.'",
   body="State its consequence, beginning with 'Consequence:'.",
   constraint="Do not use the word 'quarry'.",
   forbidden="quarr", referent="Threlkeld", format_re=r"^Consequence:"),
 dict(id="e03", antecedent="The Marchmont Fire destroyed the east wing and its contents.",
   exemplar="Example: 'Destroyed: one section of a building and what it held.'",
   body="State what it destroyed, beginning with 'Destroyed:'.",
   constraint="Do not use the phrase 'east wing'.",
   forbidden="east wing", referent="Marchmont", format_re=r"^Destroyed:"),
 dict(id="e04", antecedent="The Pentrich Assembly petitioned for shorter working hours.",
   exemplar="Example: 'Petitioned for: a reduction in the length of the working day.'",
   body="State what it petitioned for, beginning with 'Petitioned for:'.",
   constraint="Do not use the word 'hours'.",
   forbidden="hours", referent="Pentrich", format_re=r"^Petitioned for:"),
 dict(id="e05", antecedent="The Carrowmore Excavation uncovered a sequence of burial chambers.",
   exemplar="Example: 'Uncovered: a series of ancient structures.'",
   body="State what it uncovered, beginning with 'Uncovered:'.",
   constraint="Do not use the word 'burial'.",
   forbidden="burial", referent="Carrowmore", format_re=r"^Uncovered:"),
 dict(id="e06", antecedent="The Weatherby Enquiry examined the collapse of a footbridge.",
   exemplar="Example: 'Examined: the failure of a pedestrian crossing.'",
   body="State what it examined, beginning with 'Examined:'.",
   constraint="Do not use the word 'footbridge'.",
   forbidden="footbridge", referent="Weatherby", format_re=r"^Examined:"),
 dict(id="e07", antecedent="The Glenross Migration moved whole families to the southern plains.",
   exemplar="Example: 'Moved: entire households to a distant region.'",
   body="State whom it moved, beginning with 'Moved:'.",
   constraint="Do not use the word 'plains'.",
   forbidden="plains", referent="Glenross", format_re=r"^Moved:"),
 dict(id="e08", antecedent="The Ardmillan Settlement ended a claim over grazing land.",
   exemplar="Example: 'Ended: a dispute concerning use of open ground.'",
   body="State what it ended, beginning with 'Ended:'.",
   constraint="Do not use the word 'grazing'.",
   forbidden="grazing", referent="Ardmillan", format_re=r"^Ended:"),
 dict(id="e09", antecedent="The Kirkwall Census counted every household on the islands.",
   exemplar="Example: 'Counted: all domestic units across an archipelago.'",
   body="State what it counted, beginning with 'Counted:'.",
   constraint="Do not use the word 'island' or 'islands'.",
   forbidden="island", referent="Kirkwall", format_re=r"^Counted:"),
 dict(id="e10", antecedent="The Netherby Reform abolished tolls on the northern roads.",
   exemplar="Example: 'Abolished: charges levied on certain routes.'",
   body="State what it abolished, beginning with 'Abolished:'.",
   constraint="Do not use the word 'toll' or 'tolls'.",
   forbidden="toll", referent="Netherby", format_re=r"^Abolished:"),

 # ---- artefacts and objects ---------------------------------------------
 dict(id="a01", antecedent="The Saxby Astrolabe was engraved for a patron in Antwerp.",
   exemplar="Example: 'Made for: a wealthy individual in a trading city.'",
   body="State whom it was made for, beginning with 'Made for:'.",
   constraint="Do not mention Antwerp.",
   forbidden="Antwerp", referent="Saxby", format_re=r"^Made for:"),
 dict(id="a02", antecedent="The Eastholme Tapestry depicts a hunting scene in five panels.",
   exemplar="Example: 'Depicts: a pursuit rendered across several sections.'",
   body="State what it depicts, beginning with 'Depicts:'.",
   constraint="Do not use the word 'hunting'.",
   forbidden="hunting", referent="Eastholme", format_re=r"^Depicts:"),
 dict(id="a03", antecedent="The Rundle Clock strikes only at noon and at midnight.",
   exemplar="Example: 'Strikes: twice in each twenty-four hour period.'",
   body="State when it strikes, beginning with 'Strikes:'.",
   constraint="Do not use the word 'midnight'.",
   forbidden="midnight", referent="Rundle", format_re=r"^Strikes:"),
 dict(id="a04", antecedent="The Fitzhugh Chalice was buried during the civil disturbances.",
   exemplar="Example: 'Concealed during: a period of unrest.'",
   body="State when it was concealed, beginning with 'Concealed during:'.",
   constraint="Do not use the word 'buried'.",
   forbidden="buried", referent="Fitzhugh", format_re=r"^Concealed during:"),
 dict(id="a05", antecedent="The Wilbraham Organ has pipes recovered from three earlier instruments.",
   exemplar="Example: 'Contains: components salvaged from predecessors.'",
   body="State what it contains, beginning with 'Contains:'.",
   constraint="Do not use the word 'pipe' or 'pipes'.",
   forbidden="pipe", referent="Wilbraham", format_re=r"^Contains:"),
 dict(id="a06", antecedent="The Ockenden Seal authenticated grants of common pasture.",
   exemplar="Example: 'Authenticated: awards of shared land use.'",
   body="State what it authenticated, beginning with 'Authenticated:'.",
   constraint="Do not use the word 'pasture'.",
   forbidden="pasture", referent="Ockenden", format_re=r"^Authenticated:"),
 dict(id="a07", antecedent="The Merriott Bell cracked during an unusually severe frost.",
   exemplar="Example: 'Damaged by: extreme cold.'",
   body="State what damaged it, beginning with 'Damaged by:'.",
   constraint="Do not use the word 'frost'.",
   forbidden="frost", referent="Merriott", format_re=r"^Damaged by:"),
 dict(id="a08", antecedent="The Pardoe Screen separates the chancel from the nave.",
   exemplar="Example: 'Separates: two parts of a church interior.'",
   body="State what it separates, beginning with 'Separates:'.",
   constraint="Do not use the word 'chancel'.",
   forbidden="chancel", referent="Pardoe", format_re=r"^Separates:"),
 dict(id="a09", antecedent="The Hallgarth Cross marks the limit of an old sanctuary boundary.",
   exemplar="Example: 'Marks: the edge of a protected area.'",
   body="State what it marks, beginning with 'Marks:'.",
   constraint="Do not use the word 'sanctuary'.",
   forbidden="sanctuary", referent="Hallgarth", format_re=r"^Marks:"),
 dict(id="a10", antecedent="The Bewick Lantern guided boats into the harbour mouth.",
   exemplar="Example: 'Guided: small craft toward a sheltered entrance.'",
   body="State what it guided, beginning with 'Guided:'.",
   constraint="Do not use the word 'harbour'.",
   forbidden="harbour", referent="Bewick", format_re=r"^Guided:"),
]

CONDITIONS = {
    "full":          (),
    "no_constraint": ("constraints",),
    "no_exemplar":   ("exemplars",),
    "no_anchor":     ("anchors",),
    "no_connective": ("connectives",),
    "all_ablated":   ("constraints", "exemplars", "anchors", "connectives"),
}

ANCHOR_COMPONENTS = ("antecedent", "constraint", "exemplar")

print(f"{len(ITEMS)} items x {len(CONDITIONS)} conditions x 2 length regimes "
      f"= {len(ITEMS)*len(CONDITIONS)*2} forward passes")
print("  note: the 'full' condition is not padded, so full/matched and "
      "full/unmatched")
print("  are the same prompt; 60 of those passes are duplicates.")

_bad = []
for _it in ITEMS:
    _f, _r = _it["forbidden"].lower(), _it["referent"].lower()
    if _f in _r or _r in _f:                _bad.append((_it["id"], "forbidden overlaps referent"))
    if _f in _it["exemplar"].lower():       _bad.append((_it["id"], "exemplar violates constraint"))
    if _f not in _it["antecedent"].lower(): _bad.append((_it["id"], "forbidden not in antecedent"))
    if _r not in _it["antecedent"].lower(): _bad.append((_it["id"], "referent not in antecedent"))
    if _r in _it["body"].lower():           _bad.append((_it["id"], "pronoun not unbound"))
    if not re.search(r"\b(it|its)\b", _it["body"], re.I): _bad.append((_it["id"], "no pronoun"))
assert not _bad, f"ITEM SET PROBLEMS: {_bad}"
print("item set integrity: all checks passed")


from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME, dtype=DTYPE, device_map=DEVICE,
    attn_implementation="eager",   # REQUIRED: SDPA/flash return no attentions
)
model.eval()

n_layers = model.config.num_hidden_layers
n_heads  = model.config.num_attention_heads
print(f"{n_layers} layers x {n_heads} heads")
print("attn implementation:", model.config._attn_implementation)
assert model.config._attn_implementation == "eager", \
    "attention weights will be None without eager attention"

USE_CHAT_TEMPLATE = True
SYSTEM_MSG = ""          # empty: suppresses Qwen's ~20-token default preamble


def wrap(text):
    if not USE_CHAT_TEMPLATE:
        return text, 0
    msgs = ([{"role": "system", "content": SYSTEM_MSG}] if SYSTEM_MSG else []) \
           + [{"role": "user", "content": text}]
    full = tokenizer.apply_chat_template(msgs, tokenize=False,
                                         add_generation_prompt=True)
    off = full.find(text)
    if off < 0:
        raise RuntimeError("prompt text not found verbatim in templated string")
    return full, off


def tok_len(text):
    full, _ = wrap(text)
    return tokenizer(full, add_special_tokens=False)["input_ids"]


@torch.no_grad()
def attention_and_offsets(text):
    full, char_off = wrap(text)
    enc = tokenizer(full, return_tensors="pt", return_offsets_mapping=True)
    offsets = [(a - char_off, b - char_off) for a, b in
               enc.pop("offset_mapping")[0].tolist()]
    enc = {k: v.to(DEVICE) for k, v in enc.items()}
    out = model(**enc, output_attentions=True, use_cache=False)
    if out.attentions is None or out.attentions[0] is None:
        raise RuntimeError("no attention returned; check attn_implementation")
    layers = [a[0].float().cpu().numpy() for a in out.attentions]
    del out
    if DEVICE == "cuda":
        torch.cuda.empty_cache()
    return layers, offsets


@torch.no_grad()
def generate(text):
    full, _ = wrap(text)
    enc = tokenizer(full, return_tensors="pt").to(DEVICE)
    out = model.generate(**enc, max_new_tokens=MAX_NEW, do_sample=False,
                         pad_token_id=tokenizer.eos_token_id)
    return tokenizer.decode(out[0][enc["input_ids"].shape[1]:],
                            skip_special_tokens=True).strip()


def run_item(item):
    """All conditions x both length regimes for one item."""
    base_text, _ = build_prompt_with_spans(item, ())
    base_len = len(tok_len(base_text))
    rows = []
    for cond, drop in CONDITIONS.items():
        text, spans = build_prompt_with_spans(item, drop)
        for regime in ("unmatched", "matched"):
            if regime == "matched" and cond != "full":
                sent, achieved = pad_to_token_length(text, base_len, tok_len)
            else:
                sent, achieved = text, len(tok_len(text))
            sent_spans = spans
            rec = dict(item=item["id"], condition=cond, regime=regime,
                       prompt_tokens=achieved,
                       compression=base_len / max(achieved, 1))
            try:
                attn, offsets = attention_and_offsets(sent)
                ent = entropies_from_attention(attn, min_keys=MIN_KEYS)
                g = summarise_entropies(ent, by="global")
                rec.update(
                    seq=attn[0].shape[-1],
                    H1=float(g["H1"]), H2=float(g["H2"]),
                    Hnorm=float(g["Hnorm"]), span=float(g["span"]),
                    by_layer_Hnorm=[float(v) for v in
                                    summarise_entropies(ent, by="layer")["Hnorm"]],
                )
                idx = token_indices_for_spans(offsets, sent_spans,
                                              ANCHOR_COMPONENTS)
                mm = mass_measures(attn, idx, min_keys=MIN_KEYS) if idx else None
                if mm is None:
                    rec.update(mass=None, lift=None, Lambda=None,
                               n_anchor_tokens=0)
                else:
                    interior = mm["d_bound"] > 0.05
                    rec.update(
                        mass=float(mm["mass"].mean()),
                        lift=float(mm["lift"].mean()),
                        Lambda=float(mm["Lambda"].mean()),
                        n_anchor_tokens=len(idx),
                        by_layer_mass=[float(v) for v in
                                       mm["mass"].mean(axis=(1, 2))],
                        by_layer_Lambda=[float(v) for v in
                                         mm["Lambda"].mean(axis=(1, 2))],
                        # --- boundary mechanism -------------------------
                        d_bound_mean=float(mm["d_bound"].mean()),
                        d_bound_p01=float(np.quantile(mm["d_bound"], 0.01)),
                        d_bound_p05=float(np.quantile(mm["d_bound"], 0.05)),
                        d_bound_p50=float(np.quantile(mm["d_bound"], 0.50)),
                        frac_within_001=float((mm["d_bound"] < 0.01).mean()),
                        frac_within_005=float((mm["d_bound"] < 0.05).mean()),
                        Lambda_interior=(float(mm["Lambda"][interior].mean())
                                         if interior.any() else None),
                        # --- null-misspecification mechanism ------------
                        sink_mass=float(mm["sink"].mean()),
                        sink_mass_p95=float(np.quantile(mm["sink"], 0.95)),
                    )
                del attn
                out_text = generate(sent)
                sc = score_output(item, out_text)
                rec.update(output=out_text,
                           **{f"score_{k}": bool(v) for k, v in sc.items()},
                           correct=bool(overall_correct(sc)))
            except Exception as e:
                rec["error"] = repr(e)
            rows.append(rec)
    return rows


# ----------------------------------------------------------------------
# PRE-FLIGHT
# ----------------------------------------------------------------------
print("=" * 76)
print("PRE-FLIGHT")
print("=" * 76)
_probe = ITEMS[0]
_full_txt, _ = build_prompt_with_spans(_probe, ())
_g = generate(_full_txt)
_sc = score_output(_probe, _g)
print(f"  reference condition, item {_probe['id']}:")
print(f"    output: {_g[:150]!r}")
for _k, _v in _sc.items():
    print(f"    {_k:<22} {_v}")
if not all(_sc.values()):
    print("\n  WARNING: the REFERENCE condition fails. Degradation cannot be")
    print("  measured from a baseline that is already broken.")

_ratios = []
for _it in ITEMS:
    _f = len(tok_len(build_prompt_with_spans(_it, ())[0]))
    _a = len(tok_len(build_prompt_with_spans(_it, ABLATION_KEYS)[0]))
    _ratios.append(_f / max(_a, 1))
print(f"\n  achieved full-to-ablated compression: "
      f"{min(_ratios):.2f}x - {max(_ratios):.2f}x, "
      f"mean {sum(_ratios)/len(_ratios):.2f}x")
print(f"  template in use: {USE_CHAT_TEMPLATE}, system message: {SYSTEM_MSG!r}")
print("  REGISTERED range was 3.50x - 5.50x (mean 4.41x) on UNTEMPLATED")
print("  prompts. The template adds fixed tokens to both arms and so dilutes")
print("  the ratio; untemplated prompts were shown to break the reference")
print("  condition, so that range was never achievable with a sound baseline.")
print("  Report the achieved ratio as a deviation, with this cause.")
print()

# Resumable main loop.
done = set()
if RECORDS.exists():
    for line in RECORDS.read_text().splitlines():
        if line.strip():
            done.add(json.loads(line)["item"])
    print(f"resuming: {len(done)} items already recorded")

t0 = time.time()
with RECORDS.open("a") as fh:
    for k, item in enumerate(ITEMS, 1):
        if item["id"] in done:
            continue
        try:
            out = run_item(item)
            n_err = sum(1 for r in out if "error" in r)
            for row in out:
                fh.write(json.dumps(row) + "\n")
            fh.flush()
            if n_err:
                print(f"  {item['id']}: {n_err} condition(s) errored, recorded")
        except Exception as e:
            fh.write(json.dumps(dict(item=item["id"], error=repr(e))) + "\n")
            fh.flush()
            print(f"  {item['id']}: ITEM-LEVEL ERROR {e!r}")
        el = time.time() - t0
        print(f"[{k}/{len(ITEMS)}] {item['id']}  {el:6.1f}s elapsed")
        gc.collect()
        if DEVICE == "cuda":
            torch.cuda.empty_cache()

print(f"\ndone in {time.time()-t0:.1f}s -> {RECORDS}")


import collections

raw_rows = [json.loads(l) for l in RECORDS.read_text().splitlines() if l.strip()]
errors = [r for r in raw_rows if "error" in r]
rows   = [r for r in raw_rows if "error" not in r]
print(f"{len(rows)} usable records, {len(errors)} errored")
if errors:
    by_cond = collections.Counter(r.get("condition", "item-level") for r in errors)
    for k, v in by_cond.most_common():
        print(f"    {k}: {v}")
    print("  A condition that errors disproportionately makes the surviving")
    print("  sample non-random for that condition. Report these counts.")


# ----------------------------------------------------------------------
# DID THE LAYER FIX TAKE?  Check on real data, not synthetic.
# ----------------------------------------------------------------------
print()
print("=" * 76)
print("PATCH VERIFICATION ON REAL ATTENTION")
print("=" * 76)
_bl = [r["by_layer_Lambda"] for r in rows if r.get("by_layer_Lambda")]
if not _bl:
    print("  no by_layer_Lambda recorded -- cannot verify")
else:
    _v = np.array(_bl[0])
    _spread = float(np.ptp(_v))
    print(f"  layers recorded      : {len(_v)}")
    print(f"  Lambda spread across : {_spread:.4f}")
    print(f"  first three layers   : {_v[:3].round(4).tolist()}")
    print(f"  last three layers    : {_v[-3:].round(4).tolist()}")
    if _spread < 1e-9:
        print("\n  *** LAYERS ARE IDENTICAL -- THE BUG IS STILL PRESENT. ***")
        print("  Every mass-family number below is a final-layer measurement.")
        print("  Do not report any of it. Check mass_measures assigns")
        print("  mass[li, :, j] and not mass[:, :, j].")
    else:
        print("\n  layers differ: the fix is in effect, mass numbers are valid")


def pick(cond, regime, field="Hnorm"):
    return {r["item"]: r[field] for r in rows
            if r["condition"] == cond and r["regime"] == regime
            and r.get(field) is not None}


def paired_on(cond, regime, field):
    a = pick("full", "unmatched", field)
    b = pick(cond, regime, field)
    ids = sorted(set(a) & set(b))
    return ids, [a[i] for i in ids], [b[i] for i in ids]


print()
print("=" * 76)
print("H1 (PRIMARY) -- normalised entropy under ablation, LENGTH MATCHED")
print("=" * 76)
print(f"{'condition':>15} {'n':>4} {'d(H/log n)':>12} {'95% CI':>22} {'t':>8} {'dz':>7}")
print("-" * 76)
primary = {}
for cond in CONDITIONS:
    if cond == "full":
        continue
    ids, a, b = paired_on(cond, "matched", "Hnorm")
    if len(ids) < 2:
        continue
    st = paired_stats(a, b)
    primary[cond] = st
    ci = f"[{st['ci95'][0]:+.4f}, {st['ci95'][1]:+.4f}]"
    print(f"{cond:>15} {st['n']:>4} {st['mean_diff']:>+12.4f} {ci:>22} "
          f"{st['t']:>8.2f} {st['dz']:>7.2f}")
print("\n  H1 is supported for a condition if the CI excludes zero and the")
print("  difference is positive. Report every condition, not only those that")
print("  reach significance.")


print()
print("=" * 76)
print("H2 (SECONDARY) -- raw vs normalised, and the length regime")
print("=" * 76)
print(f"{'condition':>15} {'regime':>10} {'d raw H':>10} {'d H/log n':>11} "
      f"{'mean compr.':>12}")
print("-" * 76)
for cond in CONDITIONS:
    if cond == "full":
        continue
    for regime in ("unmatched", "matched"):
        ids, a1, b1 = paired_on(cond, regime, "H1")
        _,  a2, b2 = paired_on(cond, regime, "Hnorm")
        if len(ids) < 2:
            continue
        s1, s2 = paired_stats(a1, b1), paired_stats(a2, b2)
        compr = np.mean([r["compression"] for r in rows
                         if r["condition"] == cond and r["regime"] == regime])
        print(f"{cond:>15} {regime:>10} {s1['mean_diff']:>+10.4f} "
              f"{s2['mean_diff']:>+11.4f} {compr:>12.2f}x")
print("\n  H2 predicts the raw contrast is attenuated relative to the")
print("  normalised one in the UNMATCHED rows, and turns negative at")
print("  sufficient compression, while the normalised contrast stays positive")
print("  in both regimes.")


print()
print("=" * 76)
print("H2b -- THE MASS FAMILY, CONFOUNDED IN THE OPPOSITE DIRECTION")
print("=" * 76)
print(f"{'condition':>15} {'regime':>10} {'d mass':>9} {'d lift':>9} "
      f"{'d Lambda':>10} {'d Lam_int':>11}")
print("-" * 76)
for cond in CONDITIONS:
    if cond in ("full", "no_connective"):
        continue
    for regime in ("unmatched", "matched"):
        try:
            ids, a1, b1 = paired_on(cond, regime, "mass")
            _,  a2, b2 = paired_on(cond, regime, "lift")
            _,  a3, b3 = paired_on(cond, regime, "Lambda")
        except Exception:
            continue
        if len(ids) < 2:
            continue
        try:
            _, a4, b4 = paired_on(cond, regime, "Lambda_interior")
            li = f"{paired_stats(a4, b4)['mean_diff']:>+11.4f}" if len(a4) > 1 else f"{'n/a':>11}"
        except Exception:
            li = f"{'n/a':>11}"
        print(f"{cond:>15} {regime:>10} "
              f"{paired_stats(a1, b1)['mean_diff']:>+9.4f} "
              f"{paired_stats(a2, b2)['mean_diff']:>+9.3f} "
              f"{paired_stats(a3, b3)['mean_diff']:>+10.4f} {li}")
print("\n  Entropy falls under compression while mass rises; both are partly")
print("  artefact. Lift is reported only so its failure is visible. Lam_int")
print("  is Lambda over cells at least 0.05 from the unit interval's edge: if")
print("  it is stable where Lambda is not, the boundary mechanism is at work.")


print()
print("=" * 76)
print("CONTAMINATION INDEX -- scale-free, |artefact| / |effect|")
print("=" * 76)
print("  artefact = unmatched minus matched at the same ablation (length only)")
print("  effect   = matched ablated minus full (content only)")
print("  Both are in each statistic's own units, so the ratio is comparable")
print("  across statistics. Lambda being a log-odds and mass a proportion")
print("  makes a raw comparison of their distortions meaningless; this is the")
print("  comparison that is not.")
print()
print(f"{'condition':>15} {'statistic':>10} {'artefact':>10} {'effect':>10} "
      f"{'ratio':>9}")
print("-" * 76)
_cont = collections.defaultdict(list)
for cond in CONDITIONS:
    if cond in ("full", "no_connective"):
        continue
    for field in ("H1", "Hnorm", "mass", "lift", "Lambda"):
        base = pick("full", "unmatched", field)
        m = pick(cond, "matched", field)
        u = pick(cond, "unmatched", field)
        ids = sorted(set(base) & set(m) & set(u))
        if len(ids) < 2:
            continue
        artefact = float(np.mean([u[i] - m[i] for i in ids]))
        effect = float(np.mean([m[i] - base[i] for i in ids]))
        ratio = abs(artefact) / abs(effect) if effect else float("inf")
        _cont[field].append(ratio)
        print(f"{cond:>15} {field:>10} {artefact:>+10.4f} {effect:>+10.4f} "
              f"{ratio:>9.3f}")
if _cont:
    print()
    print(f"{'statistic':>10} {'median contamination':>22}")
    print("-" * 34)
    for f_, v in _cont.items():
        print(f"{f_:>10} {np.median(v):>22.3f}")
    if "Lambda" in _cont and "mass" in _cont:
        lam_, mas_ = np.median(_cont["Lambda"]), np.median(_cont["mass"])
        print(f"\n  Lambda {lam_:.3f} vs raw mass {mas_:.3f}: Lambda is "
              f"{'WORSE' if lam_ > mas_ else 'BETTER'} on a scale-free basis.")


print()
print("=" * 76)
print("MECHANISM -- why Lambda behaves as it does")
print("=" * 76)
_d = [r for r in rows if r.get("d_bound_mean") is not None]
if _d:
    print(f"  mean distance of per-cell mass to the boundary : "
          f"{np.mean([r['d_bound_mean'] for r in _d]):.4f}")
    print(f"  1st percentile                                 : "
          f"{np.mean([r['d_bound_p01'] for r in _d]):.6f}")
    print(f"  fraction of cells within 0.01 of 0 or 1        : "
          f"{np.mean([r['frac_within_001'] for r in _d]):.4f}")
    print(f"  fraction within 0.05                           : "
          f"{np.mean([r['frac_within_005'] for r in _d]):.4f}")
    print(f"  mean sink mass (first 4 key positions)         : "
          f"{np.mean([r['sink_mass'] for r in _d]):.4f}")
    print(f"  95th percentile sink mass                      : "
          f"{np.mean([r['sink_mass_p95'] for r in _d]):.4f}")
    print()
    print("  A large fraction near the boundary supports the boundary")
    print("  mechanism; a large and length-insensitive sink share supports")
    print("  null misspecification. They are not exclusive.")
else:
    print("  no boundary or sink data recorded")


print()
print("=" * 76)
print("H3 (TERTIARY) -- entropy rise against specification failure")
print("=" * 76)
acc = collections.defaultdict(list)
for r in rows:
    if "correct" in r:
        acc[(r["condition"], r["regime"])].append(r["correct"])
print(f"{'condition':>15} {'regime':>10} {'accuracy':>10} {'n':>5}")
print("-" * 45)
for (cond, regime), v in sorted(acc.items()):
    print(f"{cond:>15} {regime:>10} {np.mean(v):>10.3f} {len(v):>5}")
_base_acc = acc.get(("full", "unmatched"), [])
if _base_acc:
    _ba = float(np.mean(_base_acc))
    print(f"\n  BASELINE ACCURACY: {_ba:.3f} ({int(sum(_base_acc))}/{len(_base_acc)})")
    if _ba < 0.5:
        print("  *** FLOOR EFFECT. Below half the items succeed even when fully")
        print("  specified, so there is little room to measure degradation.")
        print("  H3 cannot be interpreted; report the floor rather than the")
        print("  correlation. The entropy and mass results are unaffected -- ")
        print("  they do not depend on the model answering correctly. ***")

xs, ys = [], []
for cond in CONDITIONS:
    if cond == "full":
        continue
    base_h = pick("full", "unmatched", "Hnorm")
    abl_h  = pick(cond, "matched", "Hnorm")
    abl_ok = pick(cond, "matched", "correct")
    for i in set(base_h) & set(abl_h) & set(abl_ok):
        xs.append(abl_h[i] - base_h[i])
        ys.append(0.0 if abl_ok[i] else 1.0)
if len(xs) > 3 and len(set(ys)) > 1:
    print(f"\n  point-biserial r(entropy rise, failure) = "
          f"{float(np.corrcoef(xs, ys)[0, 1]):+.3f} over {len(xs)} cells")
    print("  Cells are not independent across conditions within an item;")
    print("  treat as descriptive and fit a mixed model before reporting.")
else:
    print("\n  insufficient variation in correctness to estimate the association")


print()
print("=" * 76)
print("SIZING THE CONFIRMATORY STUDY FROM THIS PILOT")
print("=" * 76)
print(f"{'condition':>15} {'observed dz':>12} {'n for 80% power':>18}")
print("-" * 50)
for cond, st in primary.items():
    need = required_n(st["dz"])
    flag = "" if need <= len(ITEMS) else "   <- underpowered here"
    need_s = "n/a" if need == float("inf") else str(need)
    print(f"{cond:>15} {st['dz']:>12.3f} {need_s:>18}{flag}")
print("\n  Pilot effect sizes are upward-biased; these are not estimates.")

print()
print("=" * 76)
print("RUN COMPLETE")
print("=" * 76)
print(f"Raw data: {RECORDS}")
print("DOWNLOAD IT NOW, before the session disconnects. Everything above is")
print("reproducible from that file; nothing above is reproducible without it.")
```
